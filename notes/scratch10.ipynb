```python
from sympy import cos, sin, symbols

alpha, beta, x, y, C = symbols("alpha beta x y C")
```

```python
omega = (
    C
    / (cos(alpha) * y - sin(alpha) * x)
    / (cos(beta) * y - sin(beta) * x)
    / (
        sin((alpha + beta) / 2) * y
        + cos((alpha + beta) / 2) * x
        - cos((beta - alpha) / 2)
    )
)
omega
```

```python
print(str(omega))
```

```python
from sympy import expand, symbols

x = symbols("x")
expand((x - 3) * (x**2 - 1) * (x**2 - 4))
```

```python
from sympy import simplify

xp, yp = symbols("x' y'")
gamma = (alpha + beta) / 2
omega0 = simplify(
    omega.subs(x, -sin(gamma) * xp - cos(gamma) * yp).subs(
        y, cos(gamma) * xp - sin(gamma) * yp
    )
)
omega0
```

```python
print(str(omega0))
```

```python
omega1 = (
    (omega0 * (yp + cos((alpha - beta) / 2)))
    .subs(yp, -cos((alpha - beta) / 2))
    .factor()
)
omega1
```

```python
a, b = symbols("a b")
omega_gen = 1 / (x - b) - 1 / (x - a)
omega_gen
```

```python
omega_exp = simplify(
    omega_gen.subs(x, xp)
    .subs(b, -sin(gamma) * cos(alpha) + cos(gamma) * sin(alpha))
    .subs(a, -sin(gamma) * cos(beta) + cos(gamma) * sin(beta))
).factor()
omega_exp
```

```python
from sympy import expand_trig

omega2 = omega1.subs(C, sin(beta - alpha) * cos((beta - alpha) / 2)).factor()
omega2
```

```python
omega2.equals(omega_exp)
```

```python
omega_val = simplify(
    omega.subs(xp, x).subs(C, sin(beta - alpha) * cos((beta - alpha) / 2))
).factor()
omega_val
```

```python
simplify(denom(omega_series.coeff(dtheta, 1))).factor()
```

```python
(omega_l1 * yp).subs(yp, 0).factor()
```
