```python
from sympy import fraction

z_n, j, e = symbols("z_n j e")


def simplify_frac_j(num, denom, j, z_n, e):
    num = e_simp(z_simp(j_simp(num, j, z_n), z_n), e)
    print(num)
    denom = e_simp(z_simp(j_simp(denom, j, z_n), z_n), e)
    print(denom)
    denom = denom.expand().collect(j)
    new_denom = z_simp(j_simp(denom * denom.subs(j, -j), j, z_n), z_n)
    print(new_denom)
    new_num = z_simp(j_simp(num * denom.subs(j, -j), j, z_n), z_n)
    print(new_num)
    new_num2 = z_simp(j_simp(new_num * new_denom.subs(z_n, -1 - z_n), j, z_n), z_n)
    new_num2 = e_simp(new_num2, e)
    print(new_num2)
    new_denom2 = z_simp(j_simp(new_denom * new_denom.subs(z_n, -1 - z_n), j, z_n), z_n)
    new_denom2 = e_simp(new_denom2, e, full=False)
    print(new_denom2)
    return e_simp((new_num2 / new_denom2).expand(), e)


num, denom = fraction(expr2.factor())
n = 4
num = num.expand().subs(w, e**n * (z_n - j))
num = e_simp(num, e)
print(num)
denom = denom.expand().subs(w, e**n * (z_n - j))
denom = e_simp(denom, e)
print(denom)
simplify_frac_j(num, denom, j, z_n, e)
```

```python
from sympy import Poly, Rational
from utils import k_simp


def e_sq(expr, e, z_n):
    expr2 = e_simp(expr, e)
    u, v = symbols("u v")
    pol = Poly(expr, e)
    coeffs = pol.all_coeffs()
    print(len(coeffs))
    if len(coeffs) < 4:
        coeffs = [0] * (4 - len(coeffs)) + coeffs
    A, B, C, D = coeffs
    x = A**2 + B**2 + C**2 + D**2
    y = A * B + B * C + C * D
    z = A * C + A * D + B * D
    return x + y * (-1 - z_n) + z * z_n


def e_conj(expr, e):
    expr2 = e_simp(expr, e)
    u, v = symbols("u v")
    return (
        expr2.subs(e, u)
        .subs(e**2, v)
        .subs(e**3, e**2)
        .subs(e**4, e)
        .subs(u, e**4)
        .subs(v, e**3)
    )


def simplify_frac_k(num, denom, k, z_n):
    num = e_simp(z_simp(k_simp(num, k, z_n), z_n), e)
    print(num)
    denom = e_simp(z_simp(k_simp(denom, k, z_n), z_n), e)
    print(denom)
    denom = denom.expand().collect(k)
    new_denom = z_simp(k_simp(denom * denom.subs(k, -k), k, z_n), z_n)
    new_denom = e_simp(new_denom.subs(z_n, e**2 + e**3), e)
    print(new_denom)
    new_num = z_simp(k_simp(num * denom.subs(k, -k), k, z_n), z_n)
    # new_num = e_simp(new_num.subs(z_n, e**2 + e**3), e)
    print(new_num)
    # new_denom2 = e_simp(new_denom * e_conj(new_denom, e), e)
    new_denom2 = e_sq(new_denom, e, z_n)
    print(new_denom2)
    new_denom3 = z_simp(
        k_simp(new_denom2 * new_denom2.subs(z_n, -1 - z_n), k, z_n), z_n
    )
    print(new_denom3)
    # new_num2 = z_simp(k_simp(new_num * e_conj(new_denom, e), k, z_n), z_n)
    new_num2 = z_simp(
        k_simp(new_num * new_denom2.subs(z_n, -1 - z_n) * e_conj(new_denom, e), k, z_n),
        z_n,
    )
    # new_num2 = e_simp(new_num2.subs(z_n, e**2 + e**3), e)

    print(new_num2)
    return e_simp((new_num2 / new_denom3).subs(z_n, e**2 + e**3), e).expand()


k, e = symbols("k e")
num, denom = fraction(expr2.factor())
num = num.expand().subs(w, (-k + z_n + 2) * Rational(1, 2) * e**3)
print(num)
denom = denom.expand().subs(w, (-k + z_n + 2) * Rational(1, 2) * e**3)
print(denom)
simplify_frac_k(num, denom, k, z_n)
```
