```python
from functools import lru_cache


@lru_cache
def A(n):
    if n == 0:
        return 1
    if n == 1:
        return 7
    else:
        # print(n, 7 * A(n - 1) + A(n - 2) ** 2)
        return 7 * A(n - 1) + A(n - 2) ** 2


def Q(n):
    return Px(
        A(10 * n),
        A(10 * n + 1),
        A(10 * n + 2),
        A(10 * n + 3),
        A(10 * n + 4),
        A(10 * n + 5),
        A(10 * n + 6),
        A(10 * n + 7),
        A(10 * n + 8),
        A(10 * n + 9),
    )


def sumQ():
    return sum([Q(n) for n in range(100)])
```
